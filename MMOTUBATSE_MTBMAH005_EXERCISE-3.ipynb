```python
import cartopy.crs as ccrs
import cartopy
import cartopy.feature as cfeature
import matplotlib.pyplot as plt
```

```python
fig = plt.figure(figsize=(10, 10))
ax = plt.axes(projection=ccrs.SouthPolarStereo())

ax.set_extent([-180, 180, -90, -60], crs=ccrs.PlateCarree())

ax.add_feature(cfeature.LAND, facecolor='lightgray')
ax.add_feature(cfeature.OCEAN, facecolor='aliceblue')
ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
ax.add_feature(cfeature.BORDERS, linestyle=':')

gl = ax.gridlines(draw_labels=True,
                  linewidth=0.5,
                  color='gray',
                  alpha=0.6,
                  linestyle='--')

# Here I am annotating over Antarctica.
ax.annotate("ANTARCTICA",
            xy=(0, -90),
            xycoords=ccrs.PlateCarree()._as_mpl_transform(ax), 
            xytext=(2, 2),
            textcoords='offset points',
            fontsize=12,
            bbox=dict(boxstyle="round", fc="white", alpha=0.8),
            arrowprops=dict(arrowstyle='->', lw=0.8))

ax.annotate("SOUTHERN OCEAN",
            xy=(0, -66),
            xycoords=ccrs.PlateCarree()._as_mpl_transform(ax), 
            xytext=(2, 2),
            textcoords='offset points',
            fontsize=12,
           fontstyle='italic')

ax.text(0.95, 0.98, 'N',
        transform=ax.transAxes,
        fontsize=14,
        fontweight='bold',
        ha='center',
        va='center')

ax.arrow(0.95, 0.90, 0, 0.04,
         transform=ax.transAxes,
         width=0.003,
         head_width=0.02,
         head_length=0.03,
         color='black')

gl.xlabel_style = {'size': 10}
gl.ylabel_style = {'size': 10}

# Title
plt.title("South Polar Stereo Projection of Antarctica and the Southern Ocean",
          fontsize=12)

plt.tight_layout()
plt.show()
```

```python
cities = {
    "Walvis Bay": (14.5, -23),
    "Cape Town": (18.4, -34),
    "Rio de Janeiro": (-43.2, -23),
    "Montevideo": (-56.2, -35)
}

fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.Mercator())

ax.set_extent([-70, 30, -50, -20], crs=ccrs.PlateCarree())

ax.add_feature(cfeature.LAND, facecolor='lightgray')
ax.add_feature(cfeature.OCEAN, facecolor='aliceblue')
ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
cfeature.NaturalEarthFeature('physical', 'land', '110m')

gl = ax.gridlines(draw_labels=True,
                  linewidth=0.6,
                  color='gray',
                  alpha=0.6,
                  linestyle='--')

gl.top_labels = False
gl.right_labels = False
gl.xlabel_style = {'size': 10}
gl.ylabel_style = {'size': 10}

for city, (lon, lat) in cities.items():
    ax.plot(lon, lat,
            marker='*',
            color='black',
            markersize=5,
            transform=ccrs.PlateCarree())

    ax.text(lon + 2, lat,
            city,
            fontsize=9,
            transform=ccrs.PlateCarree())

    ax.text(0.94, 0.78, 'SOUTHERN AFRICA',
        transform=ax.transAxes,
        fontsize=10,
        ha='center',
        va='center')

    ax.text(0.10, 0.80, 'SOUTHERN SOUTH AMERICA',
        transform=ax.transAxes,
        fontsize=10,
        ha='center',
        va='center')

ax.text(0.50, 0.50, 'SOUTH ATLANTIC OCEAN',
        transform=ax.transAxes,
        fontsize=10,
        ha='center',
        va='center',
        color = 'blue')

ax.text(0.95, 0.30, 'N',
        transform=ax.transAxes,
        fontsize=14,
        fontweight='bold',
        ha='center',
        va='center')

ax.arrow(0.95, 0.20, 0, 0.04,
         transform=ax.transAxes,
         width=0.003,
         head_width=0.02,
         head_length=0.03,
         color='black')


plt.title("A classical Mercatorial representation of the South Atlantic Ocean (20°S to 50°S)",
          fontsize=14,
          weight='bold')

plt.tight_layout()
plt.show()
```
